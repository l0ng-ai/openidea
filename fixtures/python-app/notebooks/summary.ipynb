# Task summary

Counts the sample tasks by status.

```python
%load_ext autoreload
!python --version
```

```python
from collections import Counter

tasks = [("write docs", "done"), ("fix bug", "open"), ("review", "open")]
counts = Counter(status for _, status in tasks)
counts
```

```python
for status, n in sorted(counts.items()):
    print(f"{status}: {n}")
```
